```python
print("عدد الصفوف قبل التنظيف :", len(df))
```

```python

# 1) حذف الأعمدة الغير مطلوبة
cols_to_drop = [
"End_Lat", "End_Lng", "Wind_Chill(F)", "Precipitation(in)",
"Weather_Timestamp", "Airport_Code", "Description", "Country", "Turning_Loop"
]
df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])
```

```python
# 2) حذف الصفوف المكررة
before = len(df)
df = df.drop_duplicates()
print(f" تم حذف {before - len(df)} صف مكرر ")
```

```python
print("عدد NaT في Start_Time:", df["Start_Time"].isna().sum())
print("عدد NaT في End_Time:", df["End_Time"].isna().sum())
```

```python
before = len(df)
df = df.dropna(subset=["Start_Time", "End_Time"])
print(f"تم حذف {before - len(df)} صف بسبب تواريخ غير صالحة")
```

```python
df["Start_Time"] = pd.to_datetime(df["Start_Time"], errors="coerce")
df["End_Time"] = pd.to_datetime(df["End_Time"], errors="coerce")
```

```python
df["Duration_Minutes"] = (df["End_Time"] - df["Start_Time"]).dt.total_seconds() / 60
before = len(df)
df = df[(df["Duration_Minutes"] >= 0) & (df["Duration_Minutes"] <= 1440)]
print(f"تم حذف {before - len(df)} صف بسبب مدة حادث غير منطقي")
```

```python
# 5) أعمدة زمنية مساعدة
df["Year"] = df["Start_Time"].dt.year
df["Month"] = df["Start_Time"].dt.month
df["Day"] = df["Start_Time"].dt.day
df["Hour"] = df["Start_Time"].dt.hour
df["DayOfWeek"] = df["Start_Time"].dt.dayofweek
```

```python
before = len(df)
df = df[(df["Start_Lat"].between(24, 50)) & (df["Start_Lng"].between(-125, -66))]
print(f" تم حذف {before - len(df)} صف بسبب إحداثيات خارج نطاق أمريكا ")
```

```python
# 7) تعويض القيم الرقمية المفقودة بال median
numeric_cols = ["Temperature(F)", "Humidity(%)", "Pressure(in)", "Visibility(mi)", "Wind_Speed(mph)"]
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
text_cols = ["Weather_Condition", "Wind_Direction", "Sunrise_Sunset",
             "Civil_Twilight", "Nautical_Twilight", "Astronomical_Twilight",
             "Street", "Timezone", "Zipcode"]

for col in text_cols:
    df[col] = df[col].fillna("Unknown")

print("تم تعويض القيم النصية المفقودة بـ Unknown")
```

```python
for col in text_cols:
    df[col] = df[col].str.strip()

print("تم تنضيف المسافات الزيادة من الأعمدة النصية")
```

```python
remaining_nulls = df.isnull().sum()
remaining_nulls = remaining_nulls[remaining_nulls > 0]

print("أعمدة لسه فيها قيم فاضية:", remaining_nulls if len(remaining_nulls) > 0 else "مفيش! كله نضيف")
```
